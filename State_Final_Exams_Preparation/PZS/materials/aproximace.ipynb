```python
import numpy as np
import matplotlib.pyplot as plt

n = 1; t= np.linspace(0,15,1000)

aproximace_rady = np.zeros_like(t)
def funkce(n,t): return ((2/n)*(-1)**(n+1))*np.sin(n*t)
n1 = (2/n)*np.cos(n*np.pi)*np.sin(n*t)
n=2
n2=funkce(n,t)#n2 = (2/n)*np.cos(n*np.pi)*np.sin(n*t)

for n in range(1,n+1): aproximace_rady += (2/n)*np.cos(n*np.pi)*np.sin(n*t) #součet n1 a n2

plt.figure(figsize=(10,5))
plt.plot(t,n1, label="aproximace")
plt.plot(t,n2, label="aproximace2")
plt.plot(t,aproximace_rady, label="součet prvních dvou členů")

plt.xlabel("t")
plt.ylabel("Hodnota řady")
plt.legend()
plt.grid(True)
plt.show()
```
