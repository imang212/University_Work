```python
import numpy as np
import matplotlib.pyplot as plt

f_vz = 2000
freq = 5
amp = 1.0

freq = 5
T = 1/freq
numT = 50
Tmax = numT*T
n_length = int(numT*freq)
t = np.linspace(0,Tmax,n_length,endpoint=False) #mi pocatek ite periody signalu

signal = np.sin(2*np.pi*freq*t)

#sum = np.random.normal(0,1,f_vz) # li pocet ltého vzorku v rámci periody
amp = np.linspace(amp,amp*15,t.size) #měníme amplitudu v signálu
sum = np.array(amp*np.random.rand(t.size)-0.5)

sum_v_signalu = signal + sum

plt.figure(figsize=(10,5))
#plt.plot(t,signal, label='Původní signál')
plt.plot(t, sum_v_signalu, label="Signál se šumem", alpha=0.7, color='g')
plt.xlabel('Čas [s]')
plt.ylabel('Amplituda')
plt.legend()
plt.show()

amp = np.linspace(amp,amp*15,t.size) #měníme amplitudu v signálu
sum = np.array(amp*(2*np.random.rand(t.size)-1))
sum_v_signalu = signal + sum
plt.plot(t,sum_v_signalu)


#zlepšení signálu k šumu
one_per = int(n_length/numT)
periody = np.reshape(signal,[numT,one_per])
av_signal = np.sum(periody,axis=0)
av_signal /= numT
tone = np.linspace(0,T,one_per)
plt.plot(tone,av_signal)
#plt.plot(tone,signal)
```
