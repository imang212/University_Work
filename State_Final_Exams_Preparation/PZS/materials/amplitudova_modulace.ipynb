```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.signal as signal

c_f = 23
M=3
m_f=23
A0 = 1
trvani_signalu = 1
vzorkovani = 1000
#Tmax = 10*trvani_signalu

t=np.linspace(0,trvani_signalu,int(vzorkovani))

w = 2*np.pi*c_f*t
yn = np.array(A0*np.sin(2*np.pi*c_f*t))     #np.cos(2*np.pi*mod_f*t) #nosný signál f(t)=A0sin(wt)
mod_s = np.array(M*np.sin(2*np.pi*m_f*t))   #A0*np.cos(2*np.pi*mod_f*t+1) #mod signal m(t)=M*sin(qt+E) E-fazovy posun
#yt = (A0+(1 + mod_i * mod_i)*yn) #A = A0+m(t) modulovany signal y(t)=(A0+M*sin(qt+E))
f_m = np.array(A0*np.sin(w+w*np.cos(w)))   #fazova modulace

plt.figure(figsize=(18, 6))

plt.plot(t, yn, label='Nosný signál', color='g')
plt.plot(t, mod_s, label='Modulační signál', color='r')
plt.plot(t, f_m, label='AM Signal', color='b', linestyle='-')
#plt.fill_between(t, 0, yt, color='b', alpha=0.4, label='Vyplněné pole')
plt.grid(True)
plt.legend()
plt.show()

```
