```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.signal as signal

#konvoluce
alpha = 1.0
a = 6.0
t_ = np.linspace(0,2,100,endpoint=True)
h_t = np.array(alpha*np.exp(-a*t_)) #konvoluce
x_t = np.where(t_ >= 1.0,0,signal.square(2*np.pi*0.5*t_))

plt.plot(t_,h_t,label="h_t")
plt.plot(t_,x_t,label="x_t")

conv = np.convolve(x_t,h_t,mode="full")
conv = conv/np.max(conv)
t_conv = np.linspace(0,4,np.max(conv.shape))
plt.plot(t_conv,conv,label="convolution")
plt.xlabel("t")
plt.ylabel("signals")

plt.legend()
plt.show()
```
